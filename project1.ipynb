```python
from langchain.document_loaders.csv_loader import CSVLoader
loader=CSVLoader(file_path='codebasics_faqs.csv', source_column='prompt')
data=loader.load()
```

```python
data
```

```python
from langchain.prompts import PromptTemplate

prompt_template=""" given the following context and a question, generate an answer based on contentprompt_selector.
In answer try to give as much detail as possible from "response section only".
if answer is not found kindly say "I dont know." Dont try to make answer.
context: {context}
question: {question}
"""
prompt=PromptTemplate(template=prompt_template, input_variables=["context", "question"])
```

```python
from langchain.embeddings import HuggingFaceBgeEmbeddings
from langchain.vectorstores import FAISS
instructor_embedding=HuggingFaceBgeEmbeddings()
vectordb=FAISS.from_documents(documents=data, embedding=instructor_embedding)
```

```python
retrival=vectordb.as_retriever()
rdocs = retrival.invoke("I have never done programming in my life. Can I take this bootcamp?")

rdocs
```

```python
from langchain.chains import RetrievalQA
chain=RetrievalQA.from_chain_type(llm=llm, chain_type="stuff", retriever=retrival, input_key="query", return_source_documents=True)
```
